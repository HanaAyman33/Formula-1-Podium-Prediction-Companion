```python
import pandas as pd, os
data = r"D:\Formula-1-Podium-Prediction-Companion\original_data"

expected = ["circuits","constructors","constructor_results","constructor_standings",
            "drivers","driver_standings","races","results","qualifying",
            "lap_times","pit_stops","sprint_results","seasons","status"]

raw = {n: pd.read_csv(f"{data}/{n}.csv", dtype=str, keep_default_na=False, na_filter=False)
       for n in expected}

clean = {n: df.copy() for n, df in raw.items()}   # all cleaning happens on this one only
 
```

```python
def inspect(name):
    df = raw[name]
    print(f"=== {name}")
    print("shape (rows, columns):", df.shape)
    print("columns:", list(df.columns))
    print("\nfirst 3 rows:")
    print(df.head(3).to_string())
    print("\n\\N count per column:")    #null count per column
    print((df == r"\N").sum().to_string())
    print("\ntotal cells equal to \\N:", int((df == r"\N").sum().sum()))
    print("\nfully duplicated rows:", int(df.duplicated().sum()))   #checks for fully identical rows

    print("\nunique values per column:")
    for c in df.columns:
        n = df[c].nunique()
        print(f"- {c}: {n} unique", "" if n > 30 else sorted(df[c].unique()))
```

```python
def show_unique(name, col):
    print(sorted(raw[name][col].unique()))
```

```python
inspect("status")
```

```python
s = raw["status"]
print("empty or blank cells:", int((s.apply(lambda c: c.str.strip() == "")).sum().sum()))
print("statusId unique:", s["statusId"].is_unique)   #Duplicate primary keys
print("status text unique:", s["status"].is_unique)
print("statusId all digits:", s["statusId"].str.fullmatch(r"\d+").all())
print("text with extra spaces:", int((s["status"] != s["status"].str.strip()).sum()))
```

```python
show_unique("status", "status")
```

```python
inspect("seasons")   #year->primary key 
```
